```python
def war_fight(game_cards,war_players_pos,all_players):
    for pos in war_players_pos:
        if len(all_players[pos].player_deck) < 4:
            while len(all_players[pos].player_deck) != 0:
                game_cards.append(all_players[pos].remove_card())
            print(f'{all_players[pos].name} is out of the game!')
            all_players.remove(all_players[pos])
            war_players_pos.remove(pos)
        else:
            for i in range(3):
                game_cards.append(all_players[pos].remove_card())

    if len(all_players) == 1:
        return all_players
    else:
        war_cards = []
        for pos in war_players_pos:
            war_cards.append(all_players[pos].remove_card())
        game_cards = game_cards + war_cards
        war_win = highest_card(war_cards)
        if len(war_win) == 1:
            print(f'{all_players[war_win[0]].name} is the winner of this round!')
            all_players[war_win[0]].win_card(game_cards)
            return all_players
        else:
            return war_fight(game_cards, war_win, all_players)
```

```python
def highest_card(game_cards):
    card_values = []
    winners = []
    for card in game_cards:
        card_values.append(card.value)

    max_val = max(card_values)
    pos = 0
    for card in game_cards:
        if card.value == max_val:
            winners.append(pos)
        pos += 1
    return winners
```
