```python
from deck_class import Deck
from player_class import Player
from split_deck import split_deck

mydeck =Deck()
mydeck.shuffle_deck()
        

num_of_players = int(input('Enter the number of players: '))
players =[]
for i in range(num_of_players):
     name = input(f"Enter the name of player{i+1}: ")
     players.append(Player(name))

split_deck(mydeck,players)


# players[0].print_deck()
# print(len(players[0].player_deck))
```

```python
game_on = True

#while game_on:

game_cards=[]
for player in players:
    if len(player.player_deck) ==0:
        players.remove(player)
        if len(players) == 1:
            print("Game Over!")
            print(f'{players[0].name} is the winner!')
            game_on = False
            break
        else:
            continue
    else:
        game_cards.append(player.remove_card())

# get the winner pos with highest one
# do the check only one winner or many
# if one add card to him
# if now call war function (actial game cards, winnerspos, players list)
# it returns lis to players existing ones
# and then ,while game is on continue next round with the existing players!

# for card in game_cards:
#     print(card)

def highest_card(game_cards):
    card_values = []
    winners = []
    for card in game_cards:
        card_values.append(card.value)

    max_val = max(card_values)
    pos = 0
    for card in game_cards:
        if card.value == max_val:
            winners.append(pos)
        else:
            pos +=1
    return winners

winner_pos=highest_card(game_cards)
print(winner_pos)
```

```python
if len(winner_pos) == 1:
    print(f'{players[winner_pos[0]].name} is the winner of this round!')
    players[winner_pos[0]].print_deck()
    players[winner_pos[0]].win_card(game_cards)
    players[winner_pos[0]].print_deck()
```
